```python
df['Duration'].head(1)
```

```python
print(df['Duration'].isnull().sum())
```

```python
def seconds(duration_str:str):
    # Example duration_str = 26h 35m
    duration_str = duration_str.strip()
    duration_str = duration_str.split(' ')
    if len(duration_str) == 2:
        hours = duration_str[0][:-1]
        hours = int(hours)
        minutes = duration_str[1][:-1]
        minutes = int(minutes)
    elif len(duration_str) == 1:
        if 'h' in duration_str[0]:
            hours = duration_str[0][:-1]
            hours = int(hours)
            minutes = 0
        elif 'm' in duration_str[0]:
            minutes = duration_str[0][:-1]
            minutes = int(minutes)
            hours = 0
    else:
        raise ValueError("Invalid duration format")

    return hours*60*60 + minutes*60
```

```python
# Exmaple
print( seconds( '26h 35m' ) )
print( seconds( '26h 35m' ) == 26*60*60 + 35*60 )

print( seconds( '26h' ) )
print( seconds( '26h' ) == 26*60*60 + 0*60 )
```

```python
df['Duration'].apply(seconds).mean()
```

```python
avg_seconds = df['Duration'].apply(seconds).mean()
print(f"{avg_seconds:.2f}")
```

```python
df['Dep_Time'] = df['Dep_Time'].apply(lambda x: int(x[:2]))
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x: int(x[:2]))
```

```python
df.head(2)
```

```python
def time_category(value:int):
    if value < 5:
        return 'Night'
    elif value < 12:
        return 'Morning'
    elif value < 17:
        return 'Afternoon'
    elif value < 20:
        return 'Evening'
    else:
        return 'Night'
```

```python
df['Dep_Time'] = df['Dep_Time'].apply(time_category)
df['Arrival_Time'] = df['Arrival_Time'].apply(time_category)
```
